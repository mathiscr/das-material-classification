```{python}
DISTANCES = {
    "Euclidienne": euclidean_distance,
    "DTW": dtw_distance,
    "Correlation": cross_correlation_distance
}
distance_matrices = {
    "Euclidienne": distance_matrix_euclidian,
    "DTW": distance_matrix_dtw,
    "Correlation": distance_matrix_cc
}
```

```{python}
K_VALUES = range(1, 26)

results = {}

for distance_name, D in distance_matrices.items():
    accuracies = []
    for k in K_VALUES:
        scores = scores_of_classification(D, y, k)
        accuracies.append(scores.mean())
        if k == 1:
            print(scores)
    results[distance_name] = accuracies
```

```{python}
plt.figure(figsize=(10, 6))

for distance_name, accuracies in results.items():

    plt.plot(
        K_VALUES,
        accuracies,
        marker="o",
        label=distance_name
    )

plt.xlabel("Nombre de voisins (k)")
plt.ylabel("Accuracy")
plt.title("Performance du k-NN selon la distance")

plt.xticks(list(K_VALUES))
plt.ylim(0, 1)

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```
